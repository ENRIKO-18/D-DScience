# **IMPUTACIÓN**

El total de datos faltates (sin considerar "class" y "number of time pregnant") son 652 del total de datos (6912), lo que equivale a un 9.5%. Esto es un procentaje alto (>3%), por lo que creo conveniente aplicar **IMPUTACIÓN.**

```
import numpy as np

df_modif=df.copy() #Creo copia de dataframe para no ditar original
cols_with_missing = [
    'Plasma glucose concentration a 2 hours in an oral glucose tolerance test',
    'Diastolic blood pressure (mm Hg)',
    'Triceps skinfold thickness (mm)',
    '2-Hour serum insulin (mu U/ml)',
    'Body mass index (weight in kg/(height in m)^2)'
]
# Reemplazamos los ceros por NaN para poder usar funciones de imputación
df_modif[cols_with_missing] = df_modif[cols_with_missing].replace(0, np.nan)
df_modif.head(10)
```

```
# Imputación con "most_frequent" de cada columna
from sklearn.impute import SimpleImputer

imputer = SimpleImputer(strategy='most_frequent')

# IMPUTADOR solo a las columnas deseadas
df_modif[cols_with_missing] = imputer.fit_transform(df_modif[cols_with_missing])
df_modif.head(10)
```

# FILTRAR BMI

```
#Filtro "Sobrepeso" - IMC entre 25.0 y 29.9 kg/m²
df_modif=df_modif[(df_modif['Body mass index (weight in kg/(height in m)^2)']>=25.0) & (df_modif['Body mass index (weight in kg/(height in m)^2)']<=30)]
df_modif.head(10)
```

```
#Reseteo de INDEX
df_modif=df_modif.reset_index(drop=True)
df_modif.head(10)
```

```
df_modif.shape
```

# **NORMALIZACIÓN**

```
from sklearn.preprocessing import MinMaxScaler
#Creo mi lista con las columnas continuas
cont_cols = [
    'Plasma glucose concentration a 2 hours in an oral glucose tolerance test',
    'Diastolic blood pressure (mm Hg)',
    'Triceps skinfold thickness (mm)',
    '2-Hour serum insulin (mu U/ml)',
    'Body mass index (weight in kg/(height in m)^2)',
    'Diabetes pedigree function',
]
scal=MinMaxScaler()
df_modif[cont_cols]=scal.fit_transform(df_modif[cont_cols])
df_modif.head(10)
```

```
df_modif.describe()
```

# **Cambiar las variables categóricas por one-hot**

Solo la variable de salida es Categórica.

```
df_onehot=pd.get_dummies(df_modif,prefix_sep='-',columns=['Class variable (0 or 1)'])
df_onehot.head(10)
```
